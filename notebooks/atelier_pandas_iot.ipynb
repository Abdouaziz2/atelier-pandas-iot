**Partie 1 — Series**
1. Qu'est-ce qu'une Series ?

Une Series Pandas est une structure de données unidimensionnelle avec un index.

Tu peux la voir comme :

index     valeur

  0        24.5

  1        25.1


Elle possède donc principalement :

des valeurs
un index

Pandas permet de créer une Series avec :

pd.Series(...)

Avant de commencer, importe Pandas :

import pandas as pd

```python
import pandas as pd

temperatures = pd.Series([24.5, 25.1, 26.3, 27.0])

temperatures
```

**Concept :** pd.Series() crée une structure de données unidimensionnelle composée de valeurs associées à un index.

Par défaut, Pandas crée automatiquement l'index :

0, 1, 2, 3
### 2. Afficher son index

```python
temperatures.index
```

.index permet d'accéder à l'index de la Series.

Ici :

start=0 → l'index commence à 0 ;
stop=4 → s'arrête avant 4 ;
step=1 → avance de 1.
### 3. Afficher la 1ère valeur

```python
temperatures.iloc[0]
```

.iloc[] sélectionne une valeur selon sa position.

iloc[0] → première position

iloc[1] → deuxième position

### 4. Afficher la dernière valeur

```python
temperatures.iloc[-1]
```

### 5. Créer, de deux façons différentes, une Series de 4 températures avec les heures comme index

**Les heures demandées sont :**

12h, 13h, 14h, 15h

**Méthode 1 — avec index=**

```python
temperatures_heures = pd.Series(
    [24.5, 25.1, 26.3, 27.0],
    index=["12h", "13h", "14h", "15h"]
)

temperatures_heures
```

Ici, on fournit séparément :

values = [24.5, 25.1, 26.3, 27.0]
index = ["12h", "13h", "14h", "15h"]

**Méthode 2 — avec un dictionnaire**

```python
temperatures_heures = pd.Series({
    "12h": 24.5,
    "13h": 25.1,
    "14h": 26.3,
    "15h": 27.0
})

temperatures_heures
```

Ici, Pandas utilise directement :

clé        → index

valeur     → donnée

### 6. Afficher la 1ère et la dernière valeur

```python
premiere = temperatures_heures.iloc[0]
derniere = temperatures_heures.iloc[-1]

premiere, derniere
```

**Pourquoi iloc ici ?**

Même si notre index contient maintenant "12h", "13h", etc., la question demande la première et la dernière valeur, donc on travaille avec leur position :

temperatures_heures.iloc[0]   # première
temperatures_heures.iloc[-1]  # dernière

À l'inverse, si on voulait précisément la température de 12h, on pourrait utiliser le label :

temperatures_heures["12h"]

On approfondira cette différence avec loc / iloc à la question 4 de l'atelier.
